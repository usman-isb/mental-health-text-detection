### dataset 1 — sentiment analysis for mental health

source: `suchintikasarkar/sentiment-analysis-for-mental-health` on Kaggle  
classes: Anxiety, Bipolar, Depression, Normal, Personality disorder, Stress, Suicidal  
reference: [8] S. Sarkar, Kaggle, 2022

```python
target1 = RAW_DIR / 'Combined Data.csv'

if target1.exists():
    print(f'already downloaded → {target1}')
else:
    print('downloading ...')
    subprocess.run(
        ['kaggle', 'datasets', 'download',
         '-d', 'suchintikasarkar/sentiment-analysis-for-mental-health',
         '-p', str(RAW_DIR), '--unzip'],
        check=True
    )
    print('done')

df1 = pd.read_csv(target1)
print(f'rows: {len(df1):,}   columns: {list(df1.columns)}')
df1.head(3)
```
